# Notebook 01: Multimodal MRI Dataset Exploration & Validation

This notebook demonstrates loading BraTS 2021 multimodal MRI scans (T1, T1ce, T2, FLAIR) and ground truth tumor segmentation masks using the project's modular `src.data` package.

```python
import sys
import os
sys.path.insert(0, os.path.abspath('..'))

import matplotlib.pyplot as plt
from src.data.dataset import get_dataloaders
from src.data.validation import DatasetValidator

# Step 1: Run Dataset Validation Audit
validator = DatasetValidator(raw_dir='../data/raw')
report = validator.validate()
print('Dataset Validation Status:', report['status'])
```

```python
# Step 2: Load Sample Batch from PyTorch DataLoader
dataloaders = get_dataloaders(raw_dir='../data/raw', splits_dir='../data/splits', batch_size=2)
inputs, targets, slice_ids = next(iter(dataloaders['train']))

print('Input Tensor Shape (B, C, H, W):', inputs.shape)
print('Target Mask Tensor Shape (B, 1, H, W):', targets.shape)
```

```python
# Step 3: Visualize 4 MRI Modalities and Ground Truth Mask
mri = inputs[0].numpy()
mask = targets[0, 0].numpy()

fig, axes = plt.subplots(1, 5, figsize=(18, 4))
titles = ['T1', 'T1ce', 'T2', 'FLAIR', 'Tumor Mask']
for c in range(4):
    axes[c].imshow(mri[c], cmap='gray')
    axes[c].set_title(titles[c])
    axes[c].axis('off')

axes[4].imshow(mri[3], cmap='gray')
axes[4].imshow(mask, cmap='Reds', alpha=0.5)
axes[4].set_title(titles[4])
axes[4].axis('off')
plt.tight_layout()
plt.show()
```
