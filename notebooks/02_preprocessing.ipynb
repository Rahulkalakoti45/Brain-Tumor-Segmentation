# Notebook 02: Intensity Normalization & Patch Extraction Analysis

This notebook demonstrates Z-score intensity normalization, percentile clipping, and 2D spatial patch extraction.

```python
import sys
import os
sys.path.insert(0, os.path.abspath('..'))

import numpy as np
import matplotlib.pyplot as plt
from src.preprocessing.preprocessing import MRIPreprocessingPipeline
from src.preprocessing.patch_extraction import PatchExtractor
from src.data.dataset import get_dataloaders

# Load sample MRI slice
dataloaders = get_dataloaders(raw_dir='../data/raw', batch_size=1)
inputs, targets, _ = next(iter(dataloaders['train']))
raw_mri = inputs[0].numpy()
```

```python
# Apply Preprocessing Pipeline
pipeline = MRIPreprocessingPipeline(normalize=True, norm_type='zscore', clip_percentiles=(0.5, 99.5))
processed_mri = pipeline(raw_mri)

print('Raw Intensity Range:', raw_mri.min(), 'to', raw_mri.max())
print('Processed Intensity Range:', processed_mri.min(), 'to', processed_mri.max())
```

```python
# Test Patch Extraction and Reconstruction
extractor = PatchExtractor(patch_size=(64, 64), stride=(32, 32))
patches = extractor.extract_patches(processed_mri)
reconstructed = extractor.reconstruct_slice(patches, target_shape=processed_mri.shape)

print('Extracted Patches Shape:', patches.shape)
print('Reconstructed Slice Shape:', reconstructed.shape)
```
