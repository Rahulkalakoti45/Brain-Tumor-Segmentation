# Notebook 03: Hybrid Filtering (Bilateral Filter + Morphological Black Top-Hat)

This notebook visualizes the spatial noise reduction and edge sharpening achieved by combining non-linear Bilateral Filtering with morphological Black Top-Hat feature extraction.

```python
import sys
import os
sys.path.insert(0, os.path.abspath('..'))

import matplotlib.pyplot as plt
from src.data.dataset import get_dataloaders
from src.filtering.bilateral_filter import BilateralFilterModule
from src.filtering.black_tophat import BlackTopHatFilterModule
from src.filtering.hybrid_filter import HybridMRIFilter
from src.visualization.visualize import plot_filtering_comparison

dataloaders = get_dataloaders(raw_dir='../data/raw', batch_size=1)
inputs, targets, _ = next(iter(dataloaders['train']))
flair = inputs[0, 3].numpy()

b_module = BilateralFilterModule(d=5, sigma_color=75.0, sigma_space=75.0)
t_module = BlackTopHatFilterModule(kernel_size=5)
h_module = HybridMRIFilter()

smoothed = b_module.apply(flair)
tophat = t_module.apply(smoothed)
hybrid = h_module(inputs[0].numpy())[3]

plot_filtering_comparison(flair, smoothed, tophat, hybrid, save_path='../results/figures/notebook_filtering.png')
```
